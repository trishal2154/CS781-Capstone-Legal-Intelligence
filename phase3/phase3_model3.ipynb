## 4. Load Tokenizer

```python
from collections import defaultdict, Counter
import heapq
SPECIALS = ['<pad>', '<unk>', '<s>', '</s>', '<mask>']

class BytelevelBPE:

    def __init__(self, vocab_size=5000):
        self.vocab_size = vocab_size
        self.specials = SPECIALS.copy()
        self.byte_vocab = [str(i) for i in range(256)]
        self.merges = []
        self.merge_ranks = {}
        self.vocab = []
        self.stoi = {}
        self.itos = {}
        self.symbol_to_bytes = {}

    def word_to_symbols(self, word):
        byte_seq = word.encode('utf-8')
        return [str(b) for b in byte_seq] + ['</w>']

    def build_word_freqs(self, corpus):
        word_counter = Counter()
        print('Building word frequencies...')
        for text in corpus:
            if not isinstance(text, str):
                continue
            word_counter.update(text.split())
        print(f'Unique whitespace-separated words: {len(word_counter):,}')
        word_freqs = {}
        for word, freq in word_counter.items():
            symbols = tuple(self.word_to_symbols(word))
            word_freqs[symbols] = word_freqs.get(symbols, 0) + freq
        return word_freqs

    @staticmethod
    def get_pairs(symbols):
        return {(symbols[i], symbols[i + 1]) for i in range(len(symbols) - 1)}

    @staticmethod
    def pair_counts_in_word(symbols):
        counts = Counter()
        for i in range(len(symbols) - 1):
            counts[symbols[i], symbols[i + 1]] += 1
        return counts

    @staticmethod
    def merge_symbols(symbols, pair, merged_symbol):
        new_symbols = []
        i = 0
        while i < len(symbols):
            if i < len(symbols) - 1 and symbols[i] == pair[0] and (symbols[i + 1] == pair[1]):
                new_symbols.append(merged_symbol)
                i += 2
            else:
                new_symbols.append(symbols[i])
                i += 1
        return tuple(new_symbols)

    def train(self, corpus):
        word_freqs = self.build_word_freqs(corpus)
        vocab = self.specials + self.byte_vocab + ['</w>']
        seen = set(vocab)
        for b in range(256):
            self.symbol_to_bytes[str(b)] = bytes([b])
        self.symbol_to_bytes['</w>'] = b''
        words = []
        frequencies = []
        for symbols, freq in word_freqs.items():
            words.append(symbols)
            frequencies.append(freq)
        del word_freqs
        pair_counts = defaultdict(int)
        pair_to_words = defaultdict(set)
        print('Building initial pair statistics...')
        for word_id, symbols in enumerate(words):
            freq = frequencies[word_id]
            local_counts = self.pair_counts_in_word(symbols)
            for pair, count in local_counts.items():
                pair_counts[pair] += count * freq
                pair_to_words[pair].add(word_id)
        print(f'Initial unique pairs: {len(pair_counts):,}')
        heap = []
        for pair, count in pair_counts.items():
            heapq.heappush(heap, (-count, pair))
        target_merges = max(0, self.vocab_size - len(vocab))
        print(f'Target merges: {target_merges:,}')
        while len(vocab) < self.vocab_size and heap:
            while heap:
                neg_count, pair = heapq.heappop(heap)
                heap_count = -neg_count
                current_count = pair_counts.get(pair, 0)
                if current_count == heap_count and current_count > 0:
                    break
            else:
                break
            max_freq = pair_counts.get(pair, 0)
            if max_freq <= 1:
                break
            merged_symbol = f'<bpe_{len(self.merges)}>'
            left_bytes = self.symbol_to_bytes[pair[0]]
            right_bytes = self.symbol_to_bytes[pair[1]]
            self.symbol_to_bytes[merged_symbol] = left_bytes + right_bytes
            self.merges.append(pair)
            self.merge_ranks[pair] = (len(self.merges) - 1, merged_symbol)
            if merged_symbol not in seen:
                seen.add(merged_symbol)
                vocab.append(merged_symbol)
            affected_word_ids = list(pair_to_words.get(pair, set()))
            changed_pairs = set()
            for word_id in affected_word_ids:
                old_symbols = words[word_id]
                freq = frequencies[word_id]
                if pair not in self.get_pairs(old_symbols):
                    continue
                old_counts = self.pair_counts_in_word(old_symbols)
                new_symbols = self.merge_symbols(old_symbols, pair, merged_symbol)
                if new_symbols == old_symbols:
                    continue
                new_counts = self.pair_counts_in_word(new_symbols)
                words[word_id] = new_symbols
                all_pairs = set(old_counts) | set(new_counts)
                for p in all_pairs:
                    old_n = old_counts.get(p, 0)
                    new_n = new_counts.get(p, 0)
                    delta = (new_n - old_n) * freq
                    if delta != 0:
                        pair_counts[p] += delta
                        changed_pairs.add(p)
                    if new_n > 0:
                        pair_to_words[p].add(word_id)
                    else:
                        pair_to_words[p].discard(word_id)
            pair_to_words[pair].clear()
            for p in changed_pairs:
                count = pair_counts.get(p, 0)
                if count > 0:
                    heapq.heappush(heap, (-count, p))
        self.vocab = vocab
        self.vocab_size = len(vocab)
        self.stoi = {symbol: i for i, symbol in enumerate(vocab)}
        self.itos = {i: symbol for symbol, i in self.stoi.items()}
        print('\nTokenizer training complete.')
        print('Final vocabulary size:', self.vocab_size)
        print('Number of merges:', len(self.merges))

    def apply_bpe(self, word):
        symbols = self.word_to_symbols(word)
        while len(symbols) > 1:
            best_pair = None
            best_rank = float('inf')
            best_merged_symbol = None
            for i in range(len(symbols) - 1):
                pair = (symbols[i], symbols[i + 1])
                merge_info = self.merge_ranks.get(pair)
                if merge_info is not None:
                    rank, merged_symbol = merge_info
                    if rank < best_rank:
                        best_rank = rank
                        best_pair = pair
                        best_merged_symbol = merged_symbol
            if best_pair is None:
                break
            symbols = list(self.merge_symbols(tuple(symbols), best_pair, best_merged_symbol))
        return symbols

    def encode(self, text, add_bos=False, add_eos=False):
        ids = []
        if add_bos:
            ids.append(self.stoi['<s>'])
        for word in text.split():
            symbols = self.apply_bpe(word)
            for symbol in symbols:
                ids.append(self.stoi.get(symbol, self.stoi['<unk>']))
        if add_eos:
            ids.append(self.stoi['</s>'])
        return ids

    def decode(self, ids):
        words = []
        current_bytes = bytearray()
        for token_id in ids:
            symbol = self.itos.get(token_id)
            if symbol is None:
                continue
            if symbol in self.specials:
                continue
            ends_word = self._symbol_ends_word(symbol)
            byte_value = self.symbol_to_bytes.get(symbol, b'')
            current_bytes.extend(byte_value)
            if ends_word:
                try:
                    word = bytes(current_bytes).decode('utf-8')
                    words.append(word)
                except UnicodeDecodeError:
                    words.append('<unk>')
                current_bytes = bytearray()
        if current_bytes:
            try:
                words.append(bytes(current_bytes).decode('utf-8'))
            except UnicodeDecodeError:
                words.append('<unk>')
        return ' '.join(words)

    def _symbol_ends_word(self, symbol):
        if symbol == '</w>':
            return True
        for pair, (_, merged) in self.merge_ranks.items():
            if merged == symbol:
                return self._symbol_ends_word(pair[1])
        return False

    def get_vocab_size(self):
        return len(self.vocab)
```

## 9. Define The Model
Each block has gated multi-head attention and a GEGLU feed-forward layer, with residual connections. The same weights process both objectives. `causal=False` gives bidirectional encoder attention. `causal=True` gives left-to-right attention. KV caching is allowed only in causal evaluation, because adding a token in bidirectional mode would change earlier representations.


```python
import math
from dataclasses import dataclass, asdict
import torch
import torch.nn as nn
import torch.nn.functional as F


@dataclass
class HybridConfig:
    vocab_size: int
    pad_id: int
    mask_id: int
    d_model: int = 512
    n_layers: int = 12
    n_heads: int = 8
    d_ff: int = 1664
    max_seq_len: int = 512
    dropout: float = 0.1


class GatedAttention(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.n_heads = cfg.n_heads
        self.head_dim = cfg.d_model // cfg.n_heads
        self.qkv = nn.Linear(cfg.d_model, 3 * cfg.d_model, bias=False)
        self.gate = nn.Linear(cfg.d_model, cfg.d_model, bias=False)
        self.out = nn.Linear(cfg.d_model, cfg.d_model, bias=False)
        self.output_norm = nn.LayerNorm(cfg.d_model, elementwise_affine=False)
        self.dropout = nn.Dropout(cfg.dropout)

    def forward(self, x, key_valid, causal=True, past=None, use_cache=False):
        b, t, d = x.shape
        q, k, v = self.qkv(x).chunk(3, dim=-1)
        q = q.view(b, t, self.n_heads, self.head_dim).transpose(1, 2)
        k = k.view(b, t, self.n_heads, self.head_dim).transpose(1, 2)
        v = v.view(b, t, self.n_heads, self.head_dim).transpose(1, 2)
        past_len = 0 if past is None else past[0].size(2)
        if past is not None:
            k = torch.cat((past[0], k), dim=2)
            v = torch.cat((past[1], v), dim=2)

        allowed = key_valid[:, None, None, :]
        if causal:
            query_positions = torch.arange(past_len, past_len + t, device=x.device)
            key_positions = torch.arange(k.size(2), device=x.device)
            allowed = allowed & (key_positions[None, :] <= query_positions[:, None])[None, None]

        # A fully padded query row gets a harmless fallback key to avoid NaNs.
        first_key = torch.arange(k.size(2), device=x.device).eq(0)
        allowed = allowed | (~allowed.any(dim=-1, keepdim=True) & first_key)
        # Explicit mask also handles offset queries when a KV cache is present.
        context = F.scaled_dot_product_attention(
            q, k, v, attn_mask=allowed,
            dropout_p=self.dropout.p if self.training else 0.0,
            is_causal=False,
        )
        context = context.transpose(1, 2).contiguous().view(b, t, d)
        # Paper Section 2.2: GEGLU gate and non-affine output LayerNorm.
        context = context * F.gelu(self.gate(x))
        output = self.out(self.output_norm(context))
        return output, (k, v) if use_cache else None
```

```python
class GatedFeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.value = nn.Linear(cfg.d_model, cfg.d_ff, bias=False)
        self.gate = nn.Linear(cfg.d_model, cfg.d_ff, bias=False)
        self.output_norm = nn.LayerNorm(cfg.d_ff, elementwise_affine=False)
        self.out = nn.Linear(cfg.d_ff, cfg.d_model, bias=False)

    def forward(self, x):
        hidden = self.value(x) * F.gelu(self.gate(x))
        return self.out(self.output_norm(hidden))


class HybridBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.norm1 = nn.LayerNorm(cfg.d_model, elementwise_affine=False)
        self.norm2 = nn.LayerNorm(cfg.d_model, elementwise_affine=False)
        self.attn = GatedAttention(cfg)
        self.ffn = GatedFeedForward(cfg)
        self.dropout = nn.Dropout(cfg.dropout)

    def forward(self, x, valid, causal, past=None, use_cache=False):
        attention, kv = self.attn(self.norm1(x), valid, causal, past, use_cache)
        x = x + self.dropout(attention)
        x = x + self.dropout(self.ffn(self.norm2(x)))
        return x, kv


class GPTBertLM(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.cfg = cfg
        self.tok_emb = nn.Embedding(cfg.vocab_size, cfg.d_model, padding_idx=cfg.pad_id)
        self.pos_emb = nn.Embedding(cfg.max_seq_len, cfg.d_model)
        self.embedding_norm = nn.LayerNorm(cfg.d_model, elementwise_affine=False)
        self.dropout = nn.Dropout(cfg.dropout)
        self.blocks = nn.ModuleList([HybridBlock(cfg) for _ in range(cfg.n_layers)])
        self.final_norm = nn.LayerNorm(cfg.d_model, elementwise_affine=False)
        self.lm_head = nn.Linear(cfg.d_model, cfg.vocab_size, bias=False)
        self.apply(self._initialize)
        self.lm_head.weight = self.tok_emb.weight
        # Smaller residual projections keep a deep stack stable at initialization.
        for block in self.blocks:
            nn.init.normal_(block.attn.out.weight, std=0.02 / math.sqrt(2 * cfg.n_layers))
            nn.init.normal_(block.ffn.out.weight, std=0.02 / math.sqrt(2 * cfg.n_layers))
        with torch.no_grad():
            self.tok_emb.weight[cfg.pad_id].zero_()

    def _initialize(self, module):
        if isinstance(module, (nn.Linear, nn.Embedding)):
            nn.init.normal_(module.weight, std=0.02)

    def forward(self, ids, causal=True, cache=None, use_cache=False, last_only=False):
        if (cache is not None or use_cache) and (not causal or self.training):
            raise ValueError('KV caching is only used in causal evaluation mode.')
        offset = 0 if cache is None else cache['length']
        if offset + ids.size(1) > self.cfg.max_seq_len:
            raise ValueError('Context is full: rebuild from the most recent tokens.')
        positions = torch.arange(offset, offset + ids.size(1), device=ids.device)
        x = self.dropout(self.embedding_norm(self.tok_emb(ids) + self.pos_emb(positions)))
        valid = ids.ne(self.cfg.pad_id)
        if cache is not None:
            valid = torch.cat((cache['valid'], valid), dim=1)
        updated = []
        for i, block in enumerate(self.blocks):
            past = None if cache is None else cache['kv'][i]
            x, kv = block(x, valid, causal, past, use_cache)
            if use_cache:
                updated.append(kv)
        logits = self.lm_head(self.final_norm(x[:, -1:] if last_only else x))
        if use_cache:
            return logits, dict(kv=updated, valid=valid, length=offset + ids.size(1))
        return logits
```

```python
import json
import math
import random
import zipfile
from pathlib import Path
from dataclasses import asdict

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence
from sklearn.metrics import (
    precision_recall_fscore_support,
    classification_report,
)

MODEL_DIR = Path("/kaggle/input/datasets/tejasvaidya137/hybrid-model-files")

METHOD = "lora"  # "lora" or "lorma_plus"

USE_CONTEXT = True
USE_POSITION = True
USE_CLASS_WEIGHTS = True

MAX_LENGTH = 512
BATCH_SIZE = 16
EPOCHS = 20
LEARNING_RATE = 1e-4
PATIENCE = 3

RANK = 8
ALPHA = 16
SEED = 42

RUN_NAME = f"hybrid_{METHOD}_mean_eos_position"

OUTPUT_DIR = Path("/kaggle/working/phase3") / RUN_NAME
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

use_amp = device.type == "cuda"

amp_dtype = (
    torch.bfloat16
    if use_amp and torch.cuda.is_bf16_supported()
    else torch.float16
)

print("Device:", device)
print("Method:", METHOD)
print("Output:", OUTPUT_DIR)
```

```python
checkpoint = torch.load(
    MODEL_DIR / "model.pt",
    map_location="cpu",
    weights_only=True,
)

cfg = HybridConfig(**checkpoint["model_config"])

phase1_model = GPTBertLM(cfg)

phase1_model.load_state_dict(
    checkpoint["model_state_dict"],
    strict=True,
)

del checkpoint

MAX_LENGTH = min(
    MAX_LENGTH,
    cfg.max_seq_len,
)

state = load_json(MODEL_DIR / "tokenizer.json")

tokenizer = BytelevelBPE(state["vocab_size"])
tokenizer.specials = state["specials"]
tokenizer.vocab = state["vocab"]
tokenizer.vocab_size = len(state["vocab"])

tokenizer.stoi = {
    symbol: i
    for i, symbol in enumerate(tokenizer.vocab)
}

tokenizer.itos = {
    i: symbol
    for symbol, i in tokenizer.stoi.items()
}

tokenizer.merges = [
    tuple(pair) for pair in state["merges"]
]

tokenizer.merge_ranks = {
    tuple(record["pair"]): (
        record["rank"],
        record["merged_symbol"],
    )
    for record in state["merge_ranks"]
}

tokenizer.symbol_to_bytes = {
    symbol: bytes(values)
    for symbol, values in state["symbol_to_bytes"].items()
}

del state

PAD_ID = tokenizer.stoi["<pad>"]
BOS_ID = tokenizer.stoi["<s>"]
EOS_ID = tokenizer.stoi["</s>"]
UNK_ID = tokenizer.stoi["<unk>"]

if (
    len(tokenizer.vocab) != cfg.vocab_size
    or PAD_ID != cfg.pad_id
    or tokenizer.stoi["<mask>"] != cfg.mask_id
):
    raise ValueError("Tokenizer does not match this checkpoint.")

print("Maximum sequence length:", MAX_LENGTH)
```

```python
vocab = load_json(RR_DIR / "Train/label_vocab.json")

if isinstance(vocab, dict):
    for key in ("label2id", "id2label", "labels"):
        if key in vocab and isinstance(vocab[key], (dict, list)):
            vocab = vocab[key]
            break

if isinstance(vocab, list):
    label_names = vocab
elif all(str(key).isdigit() for key in vocab):
    label_names = [
        vocab[key]
        for key in sorted(vocab, key=lambda x: int(x))
    ]
else:
    label_names = sorted(
        vocab, key=lambda x: int(vocab[x])
    )

label2id = {
    label: i for i, label in enumerate(label_names)
}
id2label = dict(enumerate(label_names))
NUM_LABELS = len(label_names)


def read_documents(path, domain, labeled=True):
    raw = load_json(path)

    if isinstance(raw, dict) and "id" in raw and "text" in raw:
        records = [raw]

    elif isinstance(raw, dict):
        records = []

        for doc_id, record in raw.items():
            if isinstance(record, list):
                record = {"text": record}

            records.append({**record, "id": doc_id})

    else:
        records = raw

    documents = []
    seen = set()

    for record in records:
        doc_id = str(record["id"])
        sentences = record["text"]
        labels = record["labels"] if labeled else None

        if doc_id in seen:
            raise ValueError(f"Duplicate document ID: {doc_id}")
        seen.add(doc_id)

        if (
            not isinstance(sentences, list)
            or any(not isinstance(s, str) for s in sentences)
        ):
            raise ValueError("Expected a list of sentence strings.")

        if labeled:
            if len(sentences) != len(labels):
                raise ValueError(f"Sentence-label mismatch: {doc_id}")

            if any(label not in label2id for label in labels):
                raise ValueError(f"Unknown label in {doc_id}")

        documents.append({
            "id": doc_id,
            "domain": domain,
            "text": sentences,
            "labels": labels,
        })

    return documents


train_docs = []
dev_docs = []

for domain in ("CL", "IT"):
    train_docs += read_documents(
        RR_DIR / f"Train/{domain}_train.json",
        domain,
    )

    dev_docs += read_documents(
        RR_DIR / f"Train/{domain}_dev.json",
        domain,
    )

print("Labels:", label_names)
print("Training documents:", len(train_docs))
print("Development documents:", len(dev_docs))
```

```python
previous_prefix = tokenizer.encode("Previous sentence:")
next_prefix = tokenizer.encode("Next sentence:")
target_prefix = tokenizer.encode("Target sentence:")

overhead = (
    4
    + len(previous_prefix)
    + len(next_prefix)
    + len(target_prefix)
)

target_capacity = MAX_LENGTH - overhead

if target_capacity < 1:
    raise ValueError("MAX_LENGTH is too small.")


def make_chunks(target, left, right):
    chunks = []

    for start in range(0, len(target), target_capacity):
        part = target[start:start + target_capacity]
        available = target_capacity - len(part)

        n_left = min(len(left), available // 2)
        n_right = min(len(right), available - n_left)
        n_left = min(len(left), available - n_right)

        left_part = left[-n_left:] if n_left else []
        right_part = right[:n_right]

        ids = (
            [BOS_ID]
            + previous_prefix + left_part + [EOS_ID]
            + next_prefix + right_part + [EOS_ID]
            + target_prefix + part + [EOS_ID]
        )

        chunks.append((ids, len(part)))

    return chunks


class RRDataset(Dataset):
    def __init__(self, documents):
        self.documents = documents
        self.tokens = []
        self.index = []

        for doc_index, document in enumerate(documents):
            encoded = []

            for sentence_index, sentence in enumerate(document["text"]):
                ids = tokenizer.encode(sentence)
                encoded.append(ids or [UNK_ID])

                self.index.append(
                    (doc_index, sentence_index)
                )

            self.tokens.append(encoded)

    def __len__(self):
        return len(self.index)

    def __getitem__(self, index):
        doc_index, sentence_index = self.index[index]

        document = self.documents[doc_index]
        sentences = self.tokens[doc_index]

        left = []
        right = []

        if USE_CONTEXT:
            if sentence_index > 0:
                left = sentences[sentence_index - 1]

            if sentence_index + 1 < len(sentences):
                right = sentences[sentence_index + 1]

        chunks = make_chunks(
            sentences[sentence_index],
            left,
            right,
        )

        label = -100

        if document["labels"] is not None:
            label = label2id[
                document["labels"][sentence_index]
            ]

        position = sentence_index / max(
            1, len(sentences) - 1
        )

        return {
            "chunks": chunks,
            "label": label,
            "position": position,
            "meta": (
                document["domain"],
                document["id"],
                sentence_index,
            ),
        }


def collate_rr(examples):
    input_ids = []
    lengths = []
    owners = []
    weights = []
    positions = []

    for owner, example in enumerate(examples):
        for ids, target_length in example["chunks"]:
            input_ids.append(
                torch.tensor(ids, dtype=torch.long)
            )

            lengths.append(len(ids))
            owners.append(owner)
            weights.append(target_length)
            positions.append(example["position"])

    return {
        "input_ids": pad_sequence(
            input_ids,
            batch_first=True,
            padding_value=PAD_ID,
        ),
        "lengths": torch.tensor(lengths, dtype=torch.long),
        "owners": torch.tensor(owners, dtype=torch.long),
        "weights": torch.tensor(weights, dtype=torch.float32),
        "sentence_positions": torch.tensor(
            positions, dtype=torch.float32
        ),
        "labels": torch.tensor(
            [example["label"] for example in examples],
            dtype=torch.long,
        ),
        "meta": [example["meta"] for example in examples],
    }


train_dataset = RRDataset(train_docs)
dev_dataset = RRDataset(dev_docs)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_rr,
)

dev_loader = DataLoader(
    dev_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_rr,
)

print("Training sentences:", len(train_dataset))
print("Development sentences:", len(dev_dataset))
```

```python
class LoRA(nn.Module):
    def __init__(self, base_layer, rank, alpha):
        super().__init__()

        self.base = base_layer
        self.base.requires_grad_(False)
        self.scale = alpha / rank

        self.A = nn.Linear(
            base_layer.in_features, rank, bias=False
        )
        self.B = nn.Linear(
            rank, base_layer.out_features, bias=False
        )

        nn.init.normal_(self.A.weight, std=0.02)
        nn.init.zeros_(self.B.weight)

    def forward(self, x):
        return (
            self.base(x)
            + self.scale * self.B(self.A(x))
        )


class LoRMAPlus(nn.Module):
    def __init__(self, base_layer, rank, alpha):
        super().__init__()

        self.base = base_layer
        self.base.requires_grad_(False)
        self.scale = alpha / rank

        out_features = base_layer.out_features

        self.A = nn.Linear(
            out_features, rank, bias=False
        )
        self.B = nn.Linear(
            rank, out_features, bias=False
        )

        nn.init.normal_(self.A.weight, std=0.02)
        nn.init.zeros_(self.B.weight)

    def forward(self, x):
        original = self.base(x)
        correction = self.B(self.A(original))

        return original + self.scale * correction
```

```python
class HybridRR(nn.Module):
    def __init__(self, pretrained_model, num_labels, use_position):
        super().__init__()

        self.cfg = pretrained_model.cfg
        self.use_position = use_position

        self.tok_emb = pretrained_model.tok_emb
        self.pos_emb = pretrained_model.pos_emb
        self.embedding_norm = pretrained_model.embedding_norm
        self.embedding_dropout = pretrained_model.dropout

        self.blocks = pretrained_model.blocks
        self.final_norm = pretrained_model.final_norm

        self.dropout = nn.Dropout(self.cfg.dropout)

        feature_size = (
            2 * self.cfg.d_model + int(use_position)
        )

        self.classifier = nn.Linear(
            feature_size,
            num_labels,
        )

    def forward(
        self,
        input_ids,
        lengths,
        target_lengths,
        sentence_positions,
    ):
        length = input_ids.size(1)

        positions = torch.arange(
            length,
            device=input_ids.device,
        )

        x = (
            self.tok_emb(input_ids)
            + self.pos_emb(positions)
        )

        x = self.embedding_dropout(
            self.embedding_norm(x)
        )

        valid = input_ids.ne(self.cfg.pad_id)

        # Bidirectional attention for sentence classification
        for block in self.blocks:
            x, _ = block(
                x,
                valid,
                causal=False,
                past=None,
                use_cache=False,
            )

        x = self.final_norm(x).float()

        # Final real EOS, before right padding
        rows = torch.arange(
            x.size(0),
            device=x.device,
        )

        eos_positions = lengths - 1
        eos_vector = x[rows, eos_positions]

        # Target tokens immediately precede final EOS
        target_starts = eos_positions - target_lengths

        token_positions = positions.unsqueeze(0)

        target_mask = (
            (token_positions >= target_starts.unsqueeze(1))
            & (token_positions < eos_positions.unsqueeze(1))
            & valid
        )

        mask = target_mask.unsqueeze(-1).float()

        target_mean = (
            (x * mask).sum(dim=1)
            / mask.sum(dim=1).clamp_min(1)
        )

        features = torch.cat(
            [target_mean, eos_vector],
            dim=-1,
        )

        features = self.dropout(features)

        if self.use_position:
            features = torch.cat(
                [
                    features,
                    sentence_positions.float().unsqueeze(-1),
                ],
                dim=-1,
            )

        return self.classifier(features)
```

```python
def move_batch(batch):
    return {
        key: value.to(device) if torch.is_tensor(value) else value
        for key, value in batch.items()
    }


def get_sentence_logits(model, batch):
    chunk_logits = model(
        batch["input_ids"],
        batch["lengths"],
        batch["weights"].long(),
        batch["sentence_positions"],
    ).float()

    owners = batch["owners"]
    weights = batch["weights"]

    logits = chunk_logits.new_zeros(
        len(batch["meta"]), NUM_LABELS
    )

    logits = logits.index_add(
        0,
        owners,
        chunk_logits * weights[:, None],
    )

    totals = weights.new_zeros(len(batch["meta"]))
    totals = totals.index_add(0, owners, weights)

    return logits / totals[:, None]


counts = torch.zeros(NUM_LABELS)

for document in train_docs:
    for label in document["labels"]:
        counts[label2id[label]] += 1

class_weights = None

if USE_CLASS_WEIGHTS:
    # class_weights = counts.clamp_min(1).rsqrt()
    class_weights = counts.clamp_min(1).pow(-0.25)
    class_weights /= class_weights.mean()
    class_weights = class_weights.to(device)
    

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

print(dict(zip(label_names, counts.int().tolist())))
```

```python
def macro_scores(gold, predicted):
    precision, recall, f1, _ = precision_recall_fscore_support(
        gold,
        predicted,
        labels=list(range(NUM_LABELS)),
        average="macro",
        zero_division=0,
    )

    return {
        "precision": float(precision),
        "recall": float(recall),
        "f1": float(f1),
    }


@torch.inference_mode()
def evaluate_rr(model, loader):
    model.eval()

    gold = []
    predicted = []
    domains = []

    for batch in loader:
        batch = move_batch(batch)

        with torch.autocast(
            device_type=device.type,
            dtype=amp_dtype,
            enabled=use_amp,
        ):
            logits = get_sentence_logits(model, batch)

        gold.extend(batch["labels"].tolist())
        predicted.extend(logits.argmax(-1).tolist())
        domains.extend(meta[0] for meta in batch["meta"])

    scores = macro_scores(gold, predicted)
    scores["domains"] = {}

    for domain in ("CL", "IT"):
        indices = [
            i for i, name in enumerate(domains)
            if name == domain
        ]

        if indices:
            scores["domains"][domain] = macro_scores(
                [gold[i] for i in indices],
                [predicted[i] for i in indices],
            )

    return scores, gold, predicted
```

```python
trainable_parameters = [
    p for p in rr_model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_parameters,
    lr=LEARNING_RATE,
    weight_decay=0.01,
)

scaler = torch.amp.GradScaler(
    "cuda",
    enabled=use_amp and amp_dtype == torch.float16,
)

run_config = {
    "method": METHOD,
    "adapter_class": Adapter.__name__,
    "rank": RANK,
    "alpha": ALPHA,
    "attention_targets": ATTENTION_TARGETS,
    "ffn_targets": FFN_TARGETS,
    "max_length": MAX_LENGTH,
    "batch_size": BATCH_SIZE,
    "learning_rate": LEARNING_RATE,
    "use_context": USE_CONTEXT,
    "use_position": USE_POSITION,
    "use_class_weights": USE_CLASS_WEIGHTS,
    "pooling": "target_mean_plus_eos",
    "input_order": "previous_next_target",
    "seed": SEED,
    "backbone": "hybrid",
    "causal": False,
}

(OUTPUT_DIR / "run_config.json").write_text(
    json.dumps(run_config, indent=2),
    encoding="utf-8",
)

history = []
best_f1 = -1.0
bad_epochs = 0

for epoch in range(1, EPOCHS + 1):
    rr_model.train()

    total_loss = 0.0
    total_weight = 0.0

    for step, batch in enumerate(train_loader, 1):
        batch = move_batch(batch)
        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(
            device_type=device.type,
            dtype=amp_dtype,
            enabled=use_amp,
        ):
            logits = get_sentence_logits(rr_model, batch)
            loss = criterion(logits, batch["labels"])

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            trainable_parameters, 1.0
        )

        scaler.step(optimizer)
        scaler.update()

        weight = (
            len(batch["meta"])
            if class_weights is None
            else class_weights[batch["labels"]].sum().item()
        )

        total_loss += loss.item() * weight
        total_weight += weight

        if step % 500 == 0:
            print(
                f"Epoch {epoch} | "
                f"Batch {step}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f}"
            )

    scores, _, _ = evaluate_rr(rr_model, dev_loader)
    train_loss = total_loss / total_weight

    print(
        f"Epoch {epoch} | "
        f"Train loss: {train_loss:.4f} | "
        f"Dev Macro-F1: {scores['f1']:.4f}"
    )

    print("Domain scores:", scores["domains"])

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        **scores,
    })

    (OUTPUT_DIR / "history.json").write_text(
        json.dumps(history, indent=2),
        encoding="utf-8",
    )

    if scores["f1"] > best_f1:
        best_f1 = scores["f1"]
        bad_epochs = 0

        torch.save(
            {
                "model_state_dict": {
                    name: tensor.detach().cpu()
                    for name, tensor in rr_model.state_dict().items()
                },
                "model_config": asdict(cfg),
                "run_config": run_config,
                "label2id": label2id,
                "epoch": epoch,
                "dev_scores": scores,
            },
            OUTPUT_DIR / "best_f1.pt",
        )

        print("Saved best checkpoint.")

    else:
        bad_epochs += 1

        if bad_epochs >= PATIENCE:
            print("Early stopping.")
            break
```

```python
def get_hybrid_features(model, batch):
    input_ids = batch["input_ids"]
    lengths = batch["lengths"]
    target_lengths = batch["weights"].long()

    positions = torch.arange(
        input_ids.size(1),
        device=input_ids.device,
    )

    x = model.tok_emb(input_ids) + model.pos_emb(positions)
    x = model.embedding_dropout(model.embedding_norm(x))

    valid = input_ids.ne(model.cfg.pad_id)

    for block in model.blocks:
        x, _ = block(
            x,
            valid,
            causal=False,
            past=None,
            use_cache=False,
        )

    x = model.final_norm(x).float()

    rows = torch.arange(x.size(0), device=x.device)
    eos_positions = lengths - 1
    eos_vector = x[rows, eos_positions]

    target_starts = eos_positions - target_lengths

    target_mask = (
        (positions[None, :] >= target_starts[:, None])
        & (positions[None, :] < eos_positions[:, None])
        & valid
    )

    mask = target_mask.unsqueeze(-1).float()

    target_mean = (
        (x * mask).sum(dim=1)
        / mask.sum(dim=1).clamp_min(1)
    )

    features = torch.cat(
        [target_mean, eos_vector],
        dim=-1,
    )

    if model.use_position:
        features = torch.cat(
            [
                features,
                batch["sentence_positions"].float().unsqueeze(-1),
            ],
            dim=-1,
        )

    return features
```

```python
@torch.no_grad()
def extract_document_features(model, documents):
    model.eval()

    loader = DataLoader(
        RRDataset(documents),
        batch_size=BATCH_SIZE,
        shuffle=False,
        collate_fn=collate_rr,
    )

    feature_size = model.classifier.in_features

    document_features = {
        (doc["domain"], doc["id"]): torch.zeros(
            len(doc["text"]),
            feature_size,
        )
        for doc in documents
    }

    for step, batch in enumerate(loader, 1):
        batch = move_batch(batch)

        with torch.autocast(
            device_type=device.type,
            dtype=amp_dtype,
            enabled=use_amp,
        ):
            chunk_features = get_hybrid_features(
                model, batch
            ).float()

        owners = batch["owners"]
        weights = batch["weights"]

        features = chunk_features.new_zeros(
            len(batch["meta"]),
            feature_size,
        )

        features.index_add_(
            0,
            owners,
            chunk_features * weights[:, None],
        )

        totals = weights.new_zeros(len(batch["meta"]))
        totals.index_add_(0, owners, weights)

        features = (features / totals[:, None]).cpu()

        for meta, vector in zip(batch["meta"], features):
            domain, doc_id, sentence_index = meta

            document_features[
                (domain, doc_id)
            ][sentence_index] = vector

        if step % 100 == 0:
            print(f"Extracted {step}/{len(loader)} batches")

    records = []

    for doc in documents:
        if not doc["text"]:
            continue

        key = (doc["domain"], doc["id"])

        labels = (
            [label2id[label] for label in doc["labels"]]
            if doc["labels"] is not None
            else [-100] * len(doc["text"])
        )

        records.append({
            "features": document_features[key],
            "labels": torch.tensor(labels, dtype=torch.long),
            "key": key,
        })

    return records
```

```python
def collate_documents(records):
    return {
        "features": pad_sequence(
            [record["features"] for record in records],
            batch_first=True,
            padding_value=0.0,
        ),
        "labels": pad_sequence(
            [record["labels"] for record in records],
            batch_first=True,
            padding_value=-100,
        ),
        "lengths": torch.tensor(
            [len(record["labels"]) for record in records],
            dtype=torch.long,
        ),
        "keys": [record["key"] for record in records],
    }


doc_train_loader = DataLoader(
    train_records,
    batch_size=DOC_BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_documents,
)

doc_dev_loader = DataLoader(
    dev_records,
    batch_size=DOC_BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_documents,
)
```

```python
from torch.nn.utils.rnn import (
    pack_padded_sequence,
    pad_packed_sequence,
)


class DocumentBiGRU(nn.Module):
    def __init__(
        self,
        input_size,
        hidden_size,
        num_labels,
        dropout=0.2,
    ):
        super().__init__()

        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        self.dropout = nn.Dropout(dropout)

        self.classifier = nn.Linear(
            2 * hidden_size,
            num_labels,
        )

    def forward(self, features, lengths):
        packed = pack_padded_sequence(
            features,
            lengths.cpu(),
            batch_first=True,
            enforce_sorted=False,
        )

        packed_output, _ = self.gru(packed)

        output, _ = pad_packed_sequence(
            packed_output,
            batch_first=True,
            total_length=features.size(1),
        )

        return self.classifier(self.dropout(output))


torch.manual_seed(SEED)

doc_model = DocumentBiGRU(
    input_size=rr_model.classifier.in_features,
    hidden_size=GRU_HIDDEN,
    num_labels=NUM_LABELS,
    dropout=GRU_DROPOUT,
).to(device)
```

```python
@torch.no_grad()
def evaluate_bigru(model, loader):
    model.eval()

    gold = []
    predicted = []

    for batch in loader:
        features = batch["features"].to(device)
        labels = batch["labels"].to(device)

        logits = model(features, batch["lengths"])
        predictions = logits.argmax(dim=-1)

        valid = labels.ne(-100)

        gold.extend(labels[valid].cpu().tolist())
        predicted.extend(predictions[valid].cpu().tolist())

    return macro_scores(gold, predicted), gold, predicted
```

```python
doc_criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    ignore_index=-100,
)

doc_optimizer = torch.optim.AdamW(
    doc_model.parameters(),
    lr=GRU_LR,
    weight_decay=0.01,
)

best_doc_f1 = -1.0
bad_epochs = 0

for epoch in range(1, GRU_EPOCHS + 1):
    doc_model.train()

    total_loss = 0.0
    total_weight = 0.0

    for batch in doc_train_loader:
        features = batch["features"].to(device)
        labels = batch["labels"].to(device)

        doc_optimizer.zero_grad(set_to_none=True)

        logits = doc_model(features, batch["lengths"])

        loss = doc_criterion(
            logits.reshape(-1, NUM_LABELS),
            labels.reshape(-1),
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            doc_model.parameters(), 1.0
        )

        doc_optimizer.step()

        valid_labels = labels[labels.ne(-100)]

        weight = (
            valid_labels.numel()
            if class_weights is None
            else class_weights[valid_labels].sum().item()
        )

        total_loss += loss.item() * weight
        total_weight += weight

    scores, _, _ = evaluate_bigru(
        doc_model, doc_dev_loader
    )

    print(
        f"Epoch {epoch} | "
        f"Train loss: {total_loss / total_weight:.4f} | "
        f"Dev Macro-F1: {scores['f1']:.4f}"
    )

    if scores["f1"] > best_doc_f1:
        best_doc_f1 = scores["f1"]
        bad_epochs = 0

        torch.save(
            {
                "model_state_dict": {
                    name: tensor.detach().cpu()
                    for name, tensor in doc_model.state_dict().items()
                },
                "input_size": rr_model.classifier.in_features,
                "hidden_size": GRU_HIDDEN,
                "dropout": GRU_DROPOUT,
                "num_labels": NUM_LABELS,
                "label2id": label2id,
                "sentence_checkpoint": str(SENTENCE_CHECKPOINT),
                "use_position": rr_model.use_position,
                "pooling": "target_mean_plus_eos",
                "epoch": epoch,
                "dev_scores": scores,
            },
            BIGRU_DIR / "best_bigru.pt",
        )

        print("Saved best BiGRU checkpoint.")

    else:
        bad_epochs += 1

        if bad_epochs >= GRU_PATIENCE:
            print("Early stopping.")
            break
```

```python
test_docs = []

for domain in ("CL", "IT"):
    test_docs += read_documents(
        RR_DIR / f"Test/{domain}_test.json",
        domain,
        labeled=False,
    )

test_records = extract_document_features(
    rr_model, test_docs
)

doc_test_loader = DataLoader(
    test_records,
    batch_size=DOC_BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_documents,
)


@torch.no_grad()
def predict_bigru(model, loader, documents):
    model.eval()

    predictions = {
        (doc["domain"], doc["id"]): []
        for doc in documents
    }

    for batch in loader:
        features = batch["features"].to(device)

        logits = model(features, batch["lengths"])
        predicted_ids = logits.argmax(dim=-1).cpu()

        for row, (key, length) in enumerate(
            zip(batch["keys"], batch["lengths"].tolist())
        ):
            predictions[key] = [
                id2label[label_id]
                for label_id in predicted_ids[row, :length].tolist()
            ]

    return predictions


predictions = predict_bigru(
    doc_model,
    doc_test_loader,
    test_docs,
)

submission = {}

for (domain, doc_id), labels in predictions.items():
    if doc_id in submission:
        raise ValueError(f"Duplicate document ID: {doc_id}")

    submission[doc_id] = labels

prediction_path = BIGRU_DIR / "prediction.json"

prediction_path.write_text(
    json.dumps(submission, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

with zipfile.ZipFile(
    BIGRU_DIR / "submission.zip",
    "w",
    zipfile.ZIP_DEFLATED,
) as archive:
    archive.write(
        prediction_path,
        arcname="prediction.json",
    )

print("Saved:", BIGRU_DIR / "submission.zip")
```
